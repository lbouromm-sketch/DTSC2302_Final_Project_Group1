```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import statsmodels.api as sm
```

```python
rng = np.random.default_rng(seed = 42)
x = rng.uniform(0,10, size = 20)
noise = rng.normal(0,2, size=20)
y = 3 + 2*x +noise
n = 20

x_bar = np.mean(x) #i'm taking the mean of my x list
y_bar = np.mean(y) #i'm taking the mean of my y list

x_list = x - x_bar

y_list = y - y_bar

numerator = np.sum(x_list * y_list) 
print(numerator)

denominator = np.sum(np.square(x_list))
print(denominator)

beta_1 = numerator/denominator
beta_0 = y_bar - beta_1 * x_bar

y_hat = beta_0 + beta_1 * x

print(f"Line: y_hat = {beta_0:.2f} + {beta_1:.2f}x")   


SSE = np.sum(np.square(y - y_hat))

print(SSE)


print(f"y_hat values + {beta_0:.2f} + {beta_1:.2f}x")
```

```python
plt.scatter(x, y, color = "blue", label = "Actual Value")
plt.plot(x,y_hat, color = "red", label = f"y_hat = {beta_0:.2f} + {beta_1:.2f}x")

plt.xlabel("x")
plt.ylabel("y")

plt.title("Least Squares Regression Line")
plt.legend()
plt.show()
```

```python
MAE = 1/n * np.sum(np.abs(y - y_hat))

print(f"MAE = (1/{n}) * {np.sum(np.abs(y-y_hat)):.3f} = {MAE:.3f}")
```
